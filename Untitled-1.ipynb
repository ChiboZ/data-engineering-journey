```python
import pandas as pd

df1 = pd.DataFrame({
    'emp_id': [1, 2, 3, 4],
    'name': ['Ali', 'Sara', 'Omar', 'Lina']
})

df2 = pd.DataFrame({
    'emp_id': [3, 4, 5, 6],
    'dept': ['Sales', 'Sales', 'IT', 'HR']
})
```

```python
pd.concat([df1 , df2] , axis=1)
```

```python
pd.merge(df1, df2, on='emp_id', how='inner') 
```

```python
pd.merge(df1, df2, on='emp_id', how='left')
```
